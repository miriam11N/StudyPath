```python
import pandas as pd
import numpy as np
from pathlib import Path
from datetime import datetime, timedelta
import random

random.seed(42)
np.random.seed(42)

DATA_DIR = Path("data")
KNOWLEDGE_DIR = Path("knowledge_base")

DATA_DIR.mkdir(exist_ok=True)
KNOWLEDGE_DIR.mkdir(exist_ok=True)

print("StudyPath folders are ready.")
```

```python
courses = pd.DataFrame([
    {
        "course_id": "CS101",
        "course_name": "Introduction to Python",
        "topics": "Variables; Data Types; Loops; Functions; Lists; Dictionaries; File Handling",
        "exam_days_away": 21
    },
    {
        "course_id": "DS201",
        "course_name": "Data Science Fundamentals",
        "topics": "Data Cleaning; Pandas; Data Visualization; Statistics; Regression; Classification",
        "exam_days_away": 28
    },
    {
        "course_id": "MATH150",
        "course_name": "College Algebra",
        "topics": "Linear Equations; Functions; Polynomials; Exponents; Quadratic Equations; Graphing",
        "exam_days_away": 18
    },
    {
        "course_id": "BUS220",
        "course_name": "Business Analytics",
        "topics": "Excel; KPIs; Descriptive Statistics; Forecasting; Dashboards; Decision Making",
        "exam_days_away": 24
    }
])

courses
```

```python
first_names = [
    "Ava", "Liam", "Noah", "Emma", "Olivia",
    "Ethan", "Sophia", "Mason", "Mia", "Lucas",
    "Isabella", "James", "Amelia", "Benjamin", "Charlotte"
]

last_names = [
    "Smith", "Johnson", "Williams", "Brown", "Jones",
    "Garcia", "Miller", "Davis", "Rodriguez", "Martinez"
]

students = []

for student_id in range(1, 101):
    first_name = random.choice(first_names)
    last_name = random.choice(last_names)

    students.append({
        "student_id": f"STU{student_id:03d}",
        "student_name": f"{first_name} {last_name}",
        "email": f"{first_name.lower()}.{last_name.lower()}{student_id}@studypath.edu",
        "weekly_study_hours": random.choice([4, 5, 6, 7, 8, 10, 12, 15]),
        "preferred_session_minutes": random.choice([25, 30, 45, 60]),
        "study_style": random.choice(["Visual", "Practice-based", "Reading", "Mixed"])
    })

students = pd.DataFrame(students)

students.head()
```

```python
enrollments = []

for _, student in students.iterrows():
    selected_courses = courses.sample(
        n=random.choice([1, 2]),
        random_state=random.randint(1, 100000)
    )

    for _, course in selected_courses.iterrows():
        quiz_score = int(np.clip(np.random.normal(loc=72, scale=16), 35, 100))
        missed_sessions = random.randint(0, 4)

        if quiz_score < 60:
            risk_level = "High Risk"
        elif quiz_score < 75:
            risk_level = "Needs Support"
        else:
            risk_level = "On Track"

        enrollments.append({
            "student_id": student["student_id"],
            "course_id": course["course_id"],
            "course_name": course["course_name"],
            "exam_days_away": course["exam_days_away"],
            "weekly_study_hours": student["weekly_study_hours"],
            "preferred_session_minutes": student["preferred_session_minutes"],
            "study_style": student["study_style"],
            "latest_quiz_score": quiz_score,
            "missed_sessions": missed_sessions,
            "risk_level": risk_level
        })

enrollments = pd.DataFrame(enrollments)

enrollments.head(10)
```

```python
topic_scores = []

for _, enrollment in enrollments.iterrows():
    course_topics = courses.loc[
        courses["course_id"] == enrollment["course_id"],
        "topics"
    ].iloc[0].split(";")

    for topic in course_topics:
        score = int(np.clip(
            enrollment["latest_quiz_score"] + np.random.normal(0, 12),
            30,
            100
        ))

        if score < 60:
            mastery_level = "Weak"
        elif score < 75:
            mastery_level = "Developing"
        else:
            mastery_level = "Strong"

        topic_scores.append({
            "student_id": enrollment["student_id"],
            "course_id": enrollment["course_id"],
            "course_name": enrollment["course_name"],
            "topic": topic.strip(),
            "topic_quiz_score": score,
            "mastery_level": mastery_level
        })

topic_scores = pd.DataFrame(topic_scores)

topic_scores.head(12)
```

```python
courses.to_csv(DATA_DIR / "courses.csv", index=False)
students.to_csv(DATA_DIR / "students.csv", index=False)
enrollments.to_csv(DATA_DIR / "student_course_progress.csv", index=False)
topic_scores.to_csv(DATA_DIR / "topic_scores.csv", index=False)

print("Saved files:")
for file_path in sorted(DATA_DIR.glob("*.csv")):
    print(f"- {file_path.name}")
```

```python
def create_study_plan(student_id, course_id):
    enrollment = enrollments[
        (enrollments["student_id"] == student_id) &
        (enrollments["course_id"] == course_id)
    ].iloc[0]

    student_topics = topic_scores[
        (topic_scores["student_id"] == student_id) &
        (topic_scores["course_id"] == course_id)
    ].sort_values("topic_quiz_score")

    weak_topics = student_topics[student_topics["topic_quiz_score"] < 75]

    if weak_topics.empty:
        weak_topics = student_topics.head(3)

    weekly_minutes = enrollment["weekly_study_hours"] * 60
    sessions_per_week = max(
        3,
        weekly_minutes // enrollment["preferred_session_minutes"]
    )

    plan = []

    for index, (_, topic_row) in enumerate(weak_topics.iterrows()):
        sessions = max(1, sessions_per_week // len(weak_topics))

        plan.append({
            "priority": index + 1,
            "topic": topic_row["topic"],
            "topic_score": topic_row["topic_quiz_score"],
            "mastery_level": topic_row["mastery_level"],
            "sessions_this_week": sessions,
            "session_minutes": enrollment["preferred_session_minutes"],
            "recommended_activity": (
                "Review notes and complete practice questions"
                if topic_row["topic_quiz_score"] < 60
                else "Practice problems and self-quiz"
            )
        })

    return pd.DataFrame(plan)

example_plan = create_study_plan("STU002", "DS201")
example_plan
```

```python
def replan_after_quiz(student_id, course_id, topic, new_score):
    current_plan = create_study_plan(student_id, course_id)

    if new_score < 60:
        message = (
            f"{topic}: score {new_score}. "
            "Add 2 extra study sessions this week, review course notes, "
            "and complete a new practice quiz before moving on."
        )
    elif new_score < 75:
        message = (
            f"{topic}: score {new_score}. "
            "Keep the topic in the plan and add 1 extra practice session."
        )
    else:
        message = (
            f"{topic}: score {new_score}. "
            "This topic is improving. Keep one review session and focus more on weaker topics."
        )

    return {
        "student_id": student_id,
        "course_id": course_id,
        "topic": topic,
        "new_score": new_score,
        "agent_recommendation": message,
        "updated_plan": current_plan
    }

replan_result = replan_after_quiz(
    student_id="STU002",
    course_id="DS201",
    topic="Data Visualization",
    new_score=52
)

print(replan_result["agent_recommendation"])
replan_result["updated_plan"]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

note_documents = []
note_metadata = []

for note_file in KNOWLEDGE_DIR.glob("*.txt"):
    text = note_file.read_text(encoding="utf-8")

    for sentence in text.split("."):
        sentence = sentence.strip()

        if len(sentence) > 25:
            note_documents.append(sentence)
            note_metadata.append({
                "source_file": note_file.name,
                "text": sentence
            })

vectorizer = TfidfVectorizer(stop_words="english")
note_matrix = vectorizer.fit_transform(note_documents)

def search_course_notes(query, course_id=None, top_k=3):
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(query_vector, note_matrix).flatten()

    results = pd.DataFrame(note_metadata)
    results["similarity_score"] = scores

    if course_id:
        results = results[
            results["source_file"].str.startswith(course_id)
        ]

    return results.sort_values(
        "similarity_score",
        ascending=False
    ).head(top_k)

search_course_notes(
    query="How do I improve at data visualization?",
    course_id="DS201"
)
```

```python
def generate_practice_quiz(student_id, course_id, topic):
    note_results = search_course_notes(
        query=topic,
        course_id=course_id,
        top_k=1
    )

    context = note_results.iloc[0]["text"]

    questions = [
        {
            "question_number": 1,
            "question": f"In your own words, explain the main idea of {topic}.",
            "type": "Short answer"
        },
        {
            "question_number": 2,
            "question": f"Give one example of how {topic} could be used in {course_id}.",
            "type": "Application"
        },
        {
            "question_number": 3,
            "question": f"What is one common mistake a student could make when studying {topic}, and how can they avoid it?",
            "type": "Reflection"
        }
    ]

    return {
        "student_id": student_id,
        "course_id": course_id,
        "topic": topic,
        "retrieved_course_note": context,
        "academic_integrity_note": (
            "This is practice material for learning. "
            "Do not submit generated responses as graded coursework."
        ),
        "questions": questions
    }

practice_quiz = generate_practice_quiz(
    student_id="STU002",
    course_id="DS201",
    topic="Data Visualization"
)

print("Retrieved course note:")
print(practice_quiz["retrieved_course_note"])

print("\nPractice Questions:")
for item in practice_quiz["questions"]:
    print(f"{item['question_number']}. {item['question']} [{item['type']}]")

print("\nAcademic Integrity:")
print(practice_quiz["academic_integrity_note"])
```

```python
def get_student_summary(student_id, course_id):
    enrollment = enrollments[
        (enrollments["student_id"] == student_id) &
        (enrollments["course_id"] == course_id)
    ].iloc[0]

    student = students[
        students["student_id"] == student_id
    ].iloc[0]

    weak_topics = topic_scores[
        (topic_scores["student_id"] == student_id) &
        (topic_scores["course_id"] == course_id) &
        (topic_scores["topic_quiz_score"] < 75)
    ].sort_values("topic_quiz_score")

    return {
        "student_name": student["student_name"],
        "course_name": enrollment["course_name"],
        "exam_days_away": enrollment["exam_days_away"],
        "weekly_study_hours": enrollment["weekly_study_hours"],
        "latest_quiz_score": enrollment["latest_quiz_score"],
        "missed_sessions": enrollment["missed_sessions"],
        "risk_level": enrollment["risk_level"],
        "weak_topics": weak_topics[[
            "topic",
            "topic_quiz_score",
            "mastery_level"
        ]]
    }

summary = get_student_summary("STU002", "DS201")

print("Student:", summary["student_name"])
print("Course:", summary["course_name"])
print("Exam days away:", summary["exam_days_away"])
print("Latest quiz score:", summary["latest_quiz_score"])
print("Risk level:", summary["risk_level"])

print("\nTopics to focus on:")
summary["weak_topics"]
```
